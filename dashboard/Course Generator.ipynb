```python
import random

NUM_COURSES = 100

departments = {
    "CSE": [
        "Data Structures","Algorithms","Operating Systems","Database Systems",
        "Machine Learning","Artificial Intelligence","Cloud Computing",
        "Computer Networks","Cyber Security","Compiler Design",
        "Distributed Systems","Software Testing","Object Oriented Programming",
        "Advanced Java","Python Programming","Full Stack Development",
        "Natural Language Processing","Computer Vision","Blockchain Technology",
        "Parallel Computing","Theory of Computation","Cryptography",
        "Internet of Things","Quantum Computing","Software Architecture",
        "Data Mining","Deep Learning","Recommendation Systems",
        "Game Development","Digital Forensics"
    ],

    "IT": [
        "Web Technologies","Information Security","DevOps",
        "Software Engineering","Big Data Analytics","Enterprise Systems",
        "Mobile Computing","IT Infrastructure Management",
        "Digital Transformation","Human Computer Interaction",
        "Cloud Architecture","Data Warehousing","Business Intelligence",
        "API Development","System Integration","Agile Methodologies",
        "IT Project Management","Data Governance","IT Auditing",
        "Technology Consulting"
    ],

    "ECE": [
        "Digital Electronics","Signal Processing","Microprocessors",
        "Embedded Systems","VLSI Design","Communication Systems",
        "Wireless Networks","RF Engineering","Analog Circuits",
        "Semiconductor Devices","Antenna Theory",
        "Digital Signal Processing","Satellite Communication",
        "Optical Communication","Control Instrumentation",
        "Microwave Engineering","Radar Systems",
        "Biomedical Electronics","IoT Hardware Design",
        "Sensor Networks"
    ],

    "EEE": [
        "Power Systems","Electrical Machines","Control Systems",
        "Power Electronics","Circuit Theory",
        "Renewable Energy Systems","High Voltage Engineering",
        "Smart Grid Technology","Industrial Drives",
        "Electrical Measurements","Protection Systems",
        "Transmission Systems","Energy Management",
        "Electrical Design","Automation Engineering",
        "Power Plant Engineering","Electric Vehicles",
        "Battery Management Systems","Substation Engineering",
        "Industrial Automation"
    ],

    "ME": [
        "Thermodynamics","Fluid Mechanics","Machine Design",
        "Manufacturing Processes","Heat Transfer","CAD/CAM",
        "Robotics","Automobile Engineering","Industrial Engineering",
        "Dynamics of Machines","Finite Element Analysis",
        "Refrigeration and AC","Production Planning",
        "Mechanical Vibrations","Tool Engineering","Mechatronics",
        "Composite Materials","Welding Technology",
        "Hydraulics and Pneumatics","Plant Engineering"
    ]
}

course_ids = random.sample(range(501, 2001), NUM_COURSES)

used_courses = set()
used_course_codes = set()

print("INSERT INTO Courses (course_id, course_code, course_name, faculty_id, credits) VALUES")

for i in range(NUM_COURSES):

    # Ensure unique department-course combination
    while True:
        dept = random.choice(list(departments.keys()))
        course_name = random.choice(departments[dept])

        if (dept, course_name) not in used_courses:
            used_courses.add((dept, course_name))
            break

    # Ensure unique course code
    while True:
        course_code = f"{dept}{random.randint(100,499)}"

        if course_code not in used_course_codes:
            used_course_codes.add(course_code)
            break

    row = (
        f"({course_ids[i]}, "
        f"'{course_code}', "
        f"'{course_name}', "
        f"{random.randint(1,50)}, "
        f"{random.choice([2,3,3,4,4,5])})"
    )

    if i < NUM_COURSES - 1:
        row += ","
    else:
        row += ";"

    print(row)
```
