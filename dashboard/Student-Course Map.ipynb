```python
import pandas as pd
import random

# Load CSVs
students_df = pd.read_csv("student_id.csv")
courses_df = pd.read_csv("course_id.csv")

student_ids = students_df["student_id"].tolist()
course_ids = courses_df["course_id"].tolist()

TARGET_ROWS = 15000

rows = []
used_pairs = set()

while len(rows) < TARGET_ROWS:

    student_id = random.choice(student_ids)
    course_id = random.choice(course_ids)

    if (student_id, course_id) in used_pairs:
        continue

    used_pairs.add((student_id, course_id))
    rows.append((student_id, course_id))


# Print SQL
print("INSERT INTO Student_Course_Map (student_id, course_id) VALUES")

for i, (sid, cid) in enumerate(rows):

    line = f"({sid}, {cid})"

    if i < len(rows) - 1:
        line += ","
    else:
        line += ";"

    print(line)
```
