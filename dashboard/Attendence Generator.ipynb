```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Load mapping CSV (generated above)
mapping_df = pd.read_csv("student-course map_details.csv")

mapping_list = mapping_df[["student_id", "course_id"]].values.tolist()

TARGET_ROWS = 15000

start_date = datetime(2026, 3, 1)
num_days = 10

rows = []
used = set()

while len(rows) < TARGET_ROWS:

    student_id, course_id = random.choice(mapping_list)

    day = random.randint(0, num_days - 1)
    date = (start_date + timedelta(days=day)).strftime("%Y-%m-%d")

    key = (student_id, course_id, date)

    if key in used:
        continue

    used.add(key)

    status = random.choices(
        ["Present", "Absent"],
        weights=[0.8, 0.2]
    )[0]

    rows.append((student_id, course_id, date, status))


# Print SQL
print("INSERT INTO Attendance (student_id, course_id, attendance_date, status) VALUES")

for i, (sid, cid, date, status) in enumerate(rows):

    line = f"({sid}, {cid}, '{date}', '{status}')"

    if i < len(rows) - 1:
        line += ","
    else:
        line += ";"

    print(line)
```
