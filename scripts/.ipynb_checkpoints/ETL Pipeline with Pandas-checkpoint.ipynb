```python
import pandas as pd
import mysql.connector
import os
```

```python
BASE_PATH = r"C:\Users\KIIT0001\OneDrive\Desktop\attendence_project\data\raw"
PROCESSED_PATH = r"C:\Users\KIIT0001\OneDrive\Desktop\attendence_project\data\processed"

# Create processed folder if not exists
os.makedirs(PROCESSED_PATH, exist_ok=True)
```

```python
students = pd.read_csv(f"{BASE_PATH}/student_details.csv")
students
```

```python
courses = pd.read_csv(f"{BASE_PATH}/course_details.csv")
courses
```

```python
faculties = pd.read_csv(f"{BASE_PATH}/faculty_details.csv")
faculties
```

```python
mapping = pd.read_csv(f"{BASE_PATH}/student_course_map_details.csv")
mapping
```

```python
attendance = pd.read_csv(f"{BASE_PATH}/attendance_details.csv")
attendance
```

```python
print("CSV Loaded Successfully")
students.head()
```

```python
#Removing duplicate values
students = students.drop_duplicates()
courses = courses.drop_duplicates()
faculties = faculties.drop_duplicates()
mapping = mapping.drop_duplicates()
attendance = attendance.drop_duplicates()
```

```python
# Convert date columns
students["created_at"] = pd.to_datetime(students["created_at"])
faculties["created_at"] = pd.to_datetime(faculties["created_at"])
attendance["attendance_date"] = pd.to_datetime(attendance["attendance_date"]).dt.date
attendance["recorded_at"] = pd.to_datetime(attendance["recorded_at"])

print("Data Cleaned Successfully")
```

```python
students.to_csv(f"{PROCESSED_PATH}/clean_students.csv", index=False)
courses.to_csv(f"{PROCESSED_PATH}/clean_courses.csv", index=False)
faculties.to_csv(f"{PROCESSED_PATH}/clean_faculty.csv", index=False)
mapping.to_csv(f"{PROCESSED_PATH}/clean_mapping.csv", index=False)
attendance.to_csv(f"{PROCESSED_PATH}/clean_attendance.csv", index=False)

print("Processed CSV files saved successfully!")
```

```python
os.listdir(PROCESSED_PATH)
```

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="1234",
    database="attendance_tracking_system"
)

cursor = conn.cursor()
```

```python
# Faculty
cursor.executemany("""
INSERT IGNORE INTO faculty (faculty_id, faculty_name, email, designation, created_at)
VALUES (%s, %s, %s, %s, %s)
""",
faculties[["faculty_id", "faculty_name", "email", "designation", "created_at"]].values.tolist()
)

# Students
cursor.executemany("""
INSERT IGNORE INTO students (student_id, roll_number, full_name, email, department, semester, created_at)
VALUES (%s, %s, %s, %s, %s, %s, %s)
""",
students[["student_id", "roll_number", "full_name", "email", "department", "semester", "created_at"]].values.tolist()
)

# Courses
cursor.executemany("""
INSERT IGNORE INTO courses (course_id, course_code, course_name, faculty_id, credits)
VALUES (%s, %s, %s, %s, %s)
""",
courses[["course_id", "course_code", "course_name", "faculty_id", "credits"]].values.tolist()
)

# Mapping
cursor.executemany("""
INSERT IGNORE INTO student_course_map (student_id, course_id)
VALUES (%s, %s)
""",
mapping[["student_id", "course_id"]].values.tolist()
)

# Attendance
cursor.executemany("""
INSERT IGNORE INTO attendance (student_id, course_id, attendance_date, status, recorded_at)
VALUES (%s, %s, %s, %s, %s)
""",
attendance[["student_id", "course_id", "attendance_date", "status", "recorded_at"]].values.tolist()
)

conn.commit()
cursor.close()
conn.close()

print("Data inserted into MySQL successfully!")
```
